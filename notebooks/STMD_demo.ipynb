# STMD-inspired detection demo

This notebook explores the **competing predictive small-target motion detector** used in the PERCH proof of concept. The detector receives ordinary video frames; the right-hand DVS-like demo is only a visualization. See the repository README and `docs/METHODS.md` for the signal-flow architecture and paper links.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import Video, display
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
summary = pd.read_csv(ROOT / "results" / "results_summary.csv")
summary[["condition","diameter_px","eval_detection_fraction","eval_correct_fraction"]].round(3)
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
sub=summary[summary.condition != "no target"]
ax.plot(sub.condition, sub.eval_correct_fraction * 100, "o-", label="Correct localization")
ax.axhline(summary.iloc[0].eval_detection_fraction * 100, ls="--", label="No-target false alarms")
ax.set(xlabel="Apparent target distance", ylabel="% evaluation frames", ylim=(0,100), title="Original scene-specific benchmark")
ax.legend()
plt.show()
```
